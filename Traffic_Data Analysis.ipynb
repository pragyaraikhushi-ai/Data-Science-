STEP 2 -Import the required libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

STEP 4 - Performing EDA (Exploratory Data Analysis) o the dataset to better understand it

```python
print("Result of head() :-")
# Shows top 5 rows of the Dataset
print(data.head())
print("\n")

print("Result of tail() :-")
# Shows bottom 5 rows
print(data.tail())
print("\n")

print("Result of info() :-")
# Displays ,column names , its not null count ,data type etc 
print(data.info())
print("\n")

print("Result of describe :-")
# Displays statistical summary of the dataset
print(data.describe())
print("\n")

print("Total number of null values")
print(data.isnull().sum())
```

STEP 5 - Remove duplicate values from dataset

```python
data.duplicated()
data.drop_duplicates(inplace=True)
```

STEP 6 - Selecting rows using loc[] 

```python
data2=data.loc[0:27]
```

STEP 7 - Visualize the extracted , refined data to get insights 

```python
plt.figure(figsize=(12,6))
plt.bar(data2['State/UT/City'],data2['Total Traffic Accidents - Total'],color='orange',edgecolor="cyan")
plt.title("Number of accidental injuries in different States",color="maroon",fontsize=20)
plt.grid(color='magenta',ls=':')
plt.xlabel("States",color='green',fontsize=18)
plt.ylabel("Number of injuries",color='green',fontsize=18)
plt.tight_layout()
plt.xticks(rotation=90)
plt.show()
```

Insights :-
1) It helps to find states that have the highest accidents number.
2) It helps to design region specific safety measures.

STEP 8 - Calculate the sum of number of accidents on road , at railway crossing and in traffic

```python
road=data['Road Accidents - Total'].sum()
rail=data['Railway Crossing Accidents - Total'].sum()
traffic=data['Total Traffic Accidents - Total'].sum()
perc=np.array([road,rail,traffic])
```

STEP 9 - Creating a list of color , explosion and label for the parameters of pie chart

```python
col=['Green','purple','magenta']
exp=[0.1,0.1,0.1]
lab=['Road Accidents','Railway Crossing Accidents','Total Traffic Accidents']
```

STEP 10 - Apply subplots() function to create subplots in the figure

```python
fig , axs =plt.subplots(1,2,figsize=(14,7))
# Plots pie chart
axs[0].pie(perc,autopct='%1.1f%%',labels=lab,colors=col,explode=exp)
axs[0].set_title("Total number of accidents caused by different ways",fontsize=16,color='navy')

# Creates scatterplot
sns.scatterplot(
    ax=axs[1],
    x=data['Road Accidents - Total'],
    y=data['Total Traffic Accidents - Total'],
    color='purple'
)
axs[1].set_title("Road Accident VS Total Accidents",fontsize=16,color='coral')
axs[1].set_xlabel("Total Number of Accidents",color='indigo')
axs[1].set_ylabel("Road Accidents",color='darkgreen')
axs[1].grid(color='brown')

fig.suptitle("Accidents Occurance",fontsize=26,color='Cyan')
fig.tight_layout()
```

Insights for pie chart :-
1) Road Accidents have the highest frequency.
2) Railways Accidents are minimal.

Insights for scatter plot :-
1) They have a strong positive correlation.
2) They have a near linear relationship.
